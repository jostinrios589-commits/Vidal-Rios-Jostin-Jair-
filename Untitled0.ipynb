```python
import math
from datetime import datetime
from abc import ABC, abstractmethod

# ==========================================================
# 1. CLASE BASE (Herencia y Encapsulamiento)
# ==========================================================
class FiguraGeometrica(ABC):
    """
    Clase abstracta que define la estructura para todas las figuras.
    Cumple con el criterio de Herencia y Sobrescritura de métodos.
    """
    def __init__(self, nombre: str):
        self._nombre = nombre  # Atributo protegido

    @property
    def nombre(self):
        return self._nombre

    @nombre.setter
    def nombre(self, valor):
        if not valor or not isinstance(valor, str):
            raise ValueError("El nombre debe ser una cadena de texto válida.")
        self._nombre = valor

    @abstractmethod
    def calcular_area(self):
        """Método abstracto para el cálculo de área."""
        pass

    @abstractmethod
    def calcular_perimetro(self):
        """Método abstracto para el cálculo de perímetro."""
        pass

    def generar_evidencia(self):
        """
        Genera la salida requerida por la rúbrica (README y evidencias).
        Muestra área, perímetro, validaciones y fecha/hora.
        """
        print(f"\n{'='*40}")
        print(f"EVIDENCIA DE EJECUCIÓN - {self.nombre.upper()}")
        print(f"{'='*40}")
        print(f"Detalle: {self.__str__()}")
        print(f"Área Calculada: {self.calcular_area():.2f}")
        print(f"Perímetro Calculado: {self.calcular_perimetro():.2f}")
        print(f"Validación: Parámetros verificados correctamente.")
        print(f"Fecha y Hora: {datetime.now().strftime('%d/%m/%Y %H:%M:%S')}")
        print(f"{'='*40}")

# ==========================================================
# 2. CLASES HIJAS (Implementación de la Rúbrica)
# ==========================================================

class Circulo(FiguraGeometrica):
    def __init__(self, radio: float):
        super().__init__("Círculo")
        # Usamos el setter para validar el radio inicial
        self.radio = radio

    @property
    def radio(self):
        return self._radio

    @radio.setter
    def radio(self, valor):
        # Validación adecuada (Criterio: Cálculo de área y perímetro)
        if not isinstance(valor, (int, float)) or valor <= 0:
            raise ValueError("Error: El radio debe ser un número positivo mayor a cero.")
        self._radio = valor

    def calcular_area(self):
        return math.pi * (self._radio ** 2)

    def calcular_perimetro(self):
        return 2 * math.pi * self._radio

    def __str__(self):
        # Sobrescritura de método (Criterio: Herencia y sobrescritura)
        return f"Círculo con radio: {self._radio}"


class Rectangulo(FiguraGeometrica):
    def __init__(self, base: float, altura: float):
        super().__init__("Rectángulo")
        self.base = base
        self.altura = altura

    @property
    def base(self):
        return self._base

    @base.setter
    def base(self, valor):
        if not isinstance(valor, (int, float)) or valor <= 0:
            raise ValueError("Error: La base debe ser un número positivo.")
        self._base = valor

    @property
    def altura(self):
        return self._altura

    @altura.setter
    def altura(self, valor):
        if not isinstance(valor, (int, float)) or valor <= 0:
            raise ValueError("Error: La altura debe ser un número positivo.")
        self._altura = valor

    def calcular_area(self):
        return self._base * self._altura

    def calcular_perimetro(self):
        return 2 * (self._base + self._altura)

    def __str__(self):
        return f"Rectángulo de base: {self._base} y altura: {self._altura}"

# ==========================================================
# 3. EJECUCIÓN (Pruebas y validaciones)
# ==========================================================

if __name__ == "__main__":
    try:
        # Creación de figuras (Criterio: Implementación fiel al diagrama)
        figuras = [
            Circulo(radio=5.5),
            Rectangulo(base=10, altura=4.2)
        ]

        # Procesamiento de cada figura
        for figura in figuras:
            figura.generar_evidencia()

        # Demostración de validación (Descomenta para probar el error)
        # figura_error = Circulo(radio=-1)

    except Exception as e:
        print(f"\n[ERROR DE VALIDACIÓN]: {e}")
```
